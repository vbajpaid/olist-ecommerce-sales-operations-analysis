```python
missing_values = orders.isnull().sum()

missing_percentage = (
    missing_values / len(orders) * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing_Count": missing_values,
    "Missing_Percentage": missing_percentage
})

missing_report
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )
```

```python
orders.info()
```

```python
orders.groupby("order_status").agg(
    Total_Orders=("order_id", "count"),
    Missing_Approval=(
        "order_approved_at",
        lambda x: x.isna().sum()
    ),
    Missing_Carrier_Date=(
        "order_delivered_carrier_date",
        lambda x: x.isna().sum()
    ),
    Missing_Delivery_Date=(
        "order_delivered_customer_date",
        lambda x: x.isna().sum()
    )
).sort_values(
    by="Total_Orders",
    ascending=False
)
```

```python
print("Total rows:", len(orders))

print("Unique order IDs:", orders["order_id"].nunique())

print(
    "Duplicate order IDs:",
    orders["order_id"].duplicated().sum()
)
```

```python
orders["actual_delivery_days"] = (
    orders["order_delivered_customer_date"] 
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400
```

```python
orders[
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_delivered_customer_date",
        "actual_delivery_days"
    ]
].head(10)
```

```python
orders["estimated_delivery_days"] = (
    orders["order_estimated_delivery_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400
```

```python
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400
```

```python
delivered_orders = orders[
    (orders["order_status"] == "delivered")
    & (orders["order_delivered_customer_date"].notna())
].copy()
```

```python
print(
    "Delivered orders:",
    len(delivered_orders)
)

print(
    "Average actual delivery time:",
    round(
        delivered_orders["actual_delivery_days"].mean(),
        2
    ),
    "days"
)

print(
    "Average delivery delay:",
    round(
        delivered_orders["delivery_delay_days"].mean(),
        2
    ),
    "days"
)
```

```python
on_time_orders = (
    delivered_orders["delivery_delay_days"] <= 0
).sum()

total_delivered = len(delivered_orders)

on_time_rate = (
    on_time_orders / total_delivered * 100
)

print("On-time orders:", on_time_orders)

print("Total delivered:", total_delivered)

print("On-time delivery rate:", round(on_time_rate, 2), "%")
```

```python
print(
    "Missing actual delivery days:",
    delivered_orders["actual_delivery_days"].isna().sum()
)

print(
    "Missing delivery delay days:",
    delivered_orders["delivery_delay_days"].isna().sum()
)

print(
    "Orders delivered late:",
    (
        delivered_orders["delivery_delay_days"] > 0
    ).sum()
)
```

```python
orders["order_month"] = (
    orders["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_orders = (
    orders.groupby("order_month")
    .agg(
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

monthly_orders.head(10)
```

```python
order_items = pd.read_csv(
    r"C:\DA\ecommerce-analytics-project\data\raw\olist_order_items_dataset.csv"
)

products = pd.read_csv(
    r"C:\DA\ecommerce-analytics-project\data\raw\olist_products_dataset.csv"
)
```

```python
print("Order items:")
display(order_items.head())

print("Products:")
display(products.head())
```

```python
print("Order items shape:", order_items.shape)
print("Products shape:", products.shape)

print("\nOrder item columns:")
print(order_items.columns.tolist())

print("\nProduct columns:")
print(products.columns.tolist())
```

```python
product_sales = order_items.merge(
    products,
    on="product_id",
    how="left"
)

product_sales.head()
```

```python
category_sales = (
    product_sales.groupby("product_category_name")
    .agg(
        total_items=("order_id", "count"),
        total_sales=("price", "sum"),
        total_freight=("freight_value", "sum")
    )
    .reset_index()
)

category_sales = category_sales.sort_values(
    by="total_sales",
    ascending=False
)

category_sales.head(10)
```

```python
import matplotlib.pyplot as plt

top_10_categories = category_sales.head(10).sort_values(
    by="total_sales",
    ascending=True
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_10_categories["product_category_name"],
    top_10_categories["total_sales"]
)

plt.title("Top 10 Product Categories by Sales", fontsize=16)
plt.xlabel("Total Sales (R$)")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()
```

```python
category_sales["average_item_price"] = (
    category_sales["total_sales"]
    / category_sales["total_items"]
).round(2)

category_sales[
    [
        "product_category_name",
        "total_items",
        "total_sales",
        "average_item_price"
    ]
].head(10)
```

```python
sales_data = order_items.merge(
    orders[
        [
            "order_id",
            "order_purchase_timestamp",
            "order_status"
        ]
    ],
    on="order_id",
    how="left"
)

sales_data.head()
```

```python
delivered_sales = sales_data[
    sales_data["order_status"] == "delivered"
].copy()
```

```python
delivered_sales["order_month"] = (
    delivered_sales["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_sales = (
    delivered_sales.groupby("order_month")
    .agg(
        total_sales=("price", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

monthly_sales.head()
```

```python
import matplotlib.pyplot as plt

monthly_sales["order_month"] = (
    monthly_sales["order_month"].dt.to_timestamp()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["order_month"],
    monthly_sales["total_sales"],
    marker="o"
)

plt.title("Monthly Sales Performance")
plt.xlabel("Month")
plt.ylabel("Total Item Sales (R$)")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
print(
    "Total order items:",
    len(product_sales)
)

print(
    "Missing product categories:",
    product_sales["product_category_name"].isna().sum()
)

print(
    "Missing product category percentage:",
    round(
        product_sales["product_category_name"].isna().mean() * 100,
        2
    ),
    "%"
)
```

```python
total_sales_from_items = order_items["price"].sum()

total_sales_from_categories = category_sales["total_sales"].sum()

print(
    "Sales from order items:",
    round(total_sales_from_items, 2)
)

print(
    "Sales from category summary:",
    round(total_sales_from_categories, 2)
)

print(
    "Difference:",
    round(
        total_sales_from_items - total_sales_from_categories,
        2
    )
)
```
